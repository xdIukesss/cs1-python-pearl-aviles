```python
name = input("Enter student's name: ")
score = float(input("Enter student's score: "))

if score >= 90 and score <= 100:
    grade = "A"
elif score >= 80 and score <= 89:
    grade = "B"
elif score >= 70 and score <= 79:
    grade = "C"
else:
    grade = "F"

print("\n--- Result ---")
print("Name:", name)
print("Grade:", grade)
```
